```python
import pandas as pd
import numpy as np
import joblib

from statsmodels.tsa.seasonal import STL
from sklearn.preprocessing import MinMaxScaler
from lightgbm import LGBMRegressor
from keras.layers import LSTM, Dense, Dropout
from keras.models import Sequential
from sklearn.metrics import mean_absolute_error

data_path = 'data/taxi_pickups_area.csv'
```

```python
df = pd.read_csv(data_path)
df.shape
```

```python
cols = df.columns.to_list()

time_stamp = cols[0]
areas = cols[1:]
df[time_stamp] = pd.to_datetime(df[time_stamp])
# df.set_index(time_stamp, inplace=True)
```

```python
latest = df[time_stamp].max()

datetime_threshold = latest - pd.Timedelta(weeks=1)
datetime_threshold

train = df[df[time_stamp] < datetime_threshold]
test = df[df[time_stamp] >= datetime_threshold]

print(train.shape)
print(test.shape)
```

```python
areas = cols[1:]
cleaned_train = train.copy()


def detect_anomalies(serie, robust=True, period=672):

    stl = STL(serie, robust=robust, period=period)

    result = stl.fit()

    resid =  result.resid

    resid_m = resid.mean()
    resid_dv = resid.std()

    lower = resid_m - (3 * resid_dv)
    upper = resid_m + (3 * resid_dv)

    return serie[(resid < lower) | (resid > upper)]


for area in areas:

    serie = cleaned_train[area].copy()

    anomalies_train = detect_anomalies(serie)

    serie.loc[anomalies_train.index] = np.nan

    serie = serie.interpolate(
        method="linear",
        limit_direction="both"
    )

    cleaned_train[area] = serie
```

```python
joblib.dump(cleaned_train, 'cleaned_train.pkl')
```

```python
cleaned_train = joblib.load('cleaned_train.pkl')
cols = cleaned_train.columns.to_list()
areas = cols[1:]
time_stamp = cols[0]
cleaned_train[time_stamp] = pd.to_datetime(cleaned_train[time_stamp])
```

```python
def create_areas_lag_features(data, nlags=5, fill_value=None):

    lag_features = {}
    lags_df = pd.DataFrame()

    for lag in range(1, nlags + 1):
        lag_features[f'lag_{lag}'] = []
    
    for area in areas:


        for i, lag in enumerate(lag_features):
            lags = data[area].shift(i + 1, fill_value=fill_value)
            lag_features[lag].extend(lags.to_list())

    
    for lag in lag_features:
        lags_df[lag] = lag_features[lag]

    return lags_df
```

```python
def create_areas_rolling_feature(data, window=3):

    rolling_df = pd.DataFrame()

    for area in areas:
        for i in range(1, window + 1):
            rolling_df[f'mean_{i}'] = data[area].rolling(window=i).mean()

    return rolling_df
```

```python
def create_time_features(data, slice_size=0):

    areas_feature = []
    demands = []

    columns = ['area_number', 'month', 'day', 'hour', 'minute', 'demands']
    time_features = {
        'months': [],
        'day':[],
        'hour':[],
        'minute':[]
    }

    new_df = pd.DataFrame(columns=columns)
    for index, area in enumerate(areas):

        areas_feature.extend([index] * (len(data) - slice_size))

        time_features['months'].extend( data[time_stamp].dt.month.to_list()[slice_size:] )
        time_features['day'].extend( data[time_stamp].dt.day.to_list()[slice_size:] )
        time_features['hour'].extend( data[time_stamp].dt.hour.to_list()[slice_size:] )
        time_features['minute'].extend( data[time_stamp].dt.minute.to_list()[slice_size:] )

        demands.extend(data[area].to_list()[slice_size:]) # target variable


    new_df['area_number'] = areas_feature
    new_df['month'] = time_features['months']
    new_df['day'] = time_features['day']
    new_df['hour'] = time_features['hour']
    new_df['minute'] = time_features['minute']
    new_df['demands'] = np.array(demands)

    return new_df
```

```python
lags_df = create_areas_lag_features(cleaned_train, nlags=30)
lags_df.tail()
```

```python
rolling_df = create_areas_rolling_feature(cleaned_train, window=3)
rolling_df.tail()
```

```python
train_df = pd.concat([time_df, lags_df, rolling_df], axis=1)
train_df
```

```python
x_features = [c for c in train_df.columns if c != 'demands']
# y_train = train_df['demands']
params_small = {
    'boosting_type': 'gbdt',
    'learning_rate': 0.01,
    'n_estimators': 500,
    'num_leaves': 15,                # Drastically lower to limit tree complexity
    'max_depth': 4,                  # Very shallow trees
    'min_child_samples': 50,         # Force larger groupings to clear out noise
    'colsample_bytree': 0.7,
    'subsample': 0.7,
    'subsample_freq': 1,
    'n_jobs':-1,
    'reg_lambda': 5.0                # Strong L2 regularization to smooth out final predictions
}

model = LGBMRegressor(**params_small)

model.fit(
    train_df[x_features],
    train_df['demands']
)
```

Recursive forecasting

```python
def recursive_forecast(test_area, area_demands, nlags, window, area, steps=673):

    forecast = []

    time_features = test_area.iloc[0, 1:-1]
    lag_features = area_demands.iloc[-nlags:]

    rolling_features = []
    for w in range(1, window + 1):
        rolling_features.append(area_demands.iloc[-w:].mean())

    next_step_f = np.concatenate( (area, time_features, lag_features, rolling_features) )
    forecast.append(model.predict([next_step_f])[0])

    for step in range(1, steps):

        time_features = test_area.iloc[step, 1:-1]

        if step < nlags:
            lag_features = area_demands.iloc[-(nlags - step):]
            lag_features = np.concatenate((lag_features, np.array(forecast[-step:])))
        else:
            lag_features = np.array(forecast[-nlags:])


        rolling_features = []
        for w in range(1, window + 1):

            if step < window:
                rolling_features.append(area_demands.iloc[-(w - step):].mean())
            else:
               rolling_features.append(np.array(forecast[-step:]).sum() / step)

        next_step_f = np.concatenate( (area, time_features, lag_features, rolling_features) )
        forecast.append(model.predict([next_step_f])[0])


    return np.array(forecast)
```

```python
test_df = create_time_features(test)
area_numbers = train_df['area_number'].unique()
areas_forecast = {k:None for k in area_numbers}

n_lags = 30
window = 3

for nb in area_numbers:

    test_area = test_df[test_df['area_number'] == nb]
    area_demands = train_df[ train_df['area_number'] == nb ]['demands']

    areas_forecast[nb] = recursive_forecast(test_area, area_demands, n_lags, window, np.array([nb]), len(test_area))

test_df['pred'] = None

for nb in areas_forecast:
    test_df.loc[test_df['area_number'] == nb, 'pred'] = areas_forecast[nb]


print("mean absolute error : ", mean_absolute_error(y_true=test_df['demands'], y_pred=test_df['pred']))
```

feature engineer for sequential models

```python
def create_sequences(data: pd.Series, seq_len=96):

    X = []
    i = seq_len

    npv = data.values.copy()

    while i < len(npv):
        X.append(npv[i-seq_len : i])
        i += 1

    return X
```

```python
areas_time_steps = []
for area in areas:
    areas_time_steps.extend(create_sequences(cleaned_train[area]))

areas_time_steps = np.array(areas_time_steps)
areas_time_steps.shape
```

```python
slice_size = 96
df_train = create_time_features(cleaned_train, slice_size=slice_size)
df_test = create_time_features(test, slice_size=slice_size)

df_train
```

Min Max Scaling for sequence and target variable

```python
seq_scaler = MinMaxScaler(feature_range=(0, 1))
y_scaler = MinMaxScaler(feature_range=(0, 1))

# df_train
y_train = df_train['demands'].to_numpy()

# # y_train
areas_time_steps = seq_scaler.fit_transform(areas_time_steps)
y_train = y_scaler.fit_transform(y_train.reshape(-1, 1))

X_train = areas_time_steps[:, :, None] * df_train.to_numpy()[:, None, :]
X_train.shape
```

create and compile LSTM model

```python
model = Sequential([

    LSTM(units=32, return_sequences=True, input_shape=X_train.shape[1:]),
    Dropout(0.2),

    LSTM(units=16, return_sequences=False),

    Dense(units=1)
])

model.compile(
    optimizer='adam',
    loss='mean_squared_error',
    metrics=['mean_absolute_error']
)
```

Fit LSTM model

```python
history = model.fit(
    X_train,
    y_train,
    epochs=4,
    batch_size=128,
    verbose=1
)
```

save LSTM model

```python
model.save("lstm_areas_ts.keras")
```
